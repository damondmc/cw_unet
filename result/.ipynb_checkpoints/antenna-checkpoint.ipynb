```python
import numpy as np
from tqdm import tqdm 
import matplotlib.pyplot as plt

def geometric_factor(A, B, C, psi, cosi):
    cos2p, sin2p = np.cos(2*psi)**2, np.sin(2*psi)**2
    a1 = 0.25 * (1+cosi**2)**2 * cos2p**2 + cosi**2 * sin2p**2
    a2 = 0.25 * (1+cosi**2)**2 * sin2p**2 + cosi**2 * cos2p**2
    a3 = 0.25 * (1-cosi**2)**2 * sin2p * cos2p    
    return 25./4. * ( a1*A + a2*B + 2*a3*C )
```

```python
det = 'L1'
alpha, delta, A, B, C, D = np.loadtxt(f'{det}_antennamatrix.txt', comments='%%').T

# Load RA and Dec from all seeds
seed = 0
size = (512, 64)
Tsft = 14400
f0 = 20

filename = f'/scratch/kriles_root/kriles0/damoncht/unet_f/data/validation/skymap_{f0}Hz_H1L1_{size[0]}x{size[1]}_{Tsft}s_4c_traindata_n1000_seed{seed}.npz'
data = np.load(filename, allow_pickle=True)
params = data['params']
psi = params[:200, -4]
cosi = params[:200, -3]

R2 = []
for i in range(2000):
    res = geometric_factor(A[i], B[i], C[i], psi, cosi)
    R2.append(res)
    
R2 = np.concatenate(R2)

with open(f"{det}_geofactor.txt", "w") as f:
    for r2 in R2:
        f.write("{}\n".format(r2))
```
