```python
import numpy as np
from genData import * 
import time
```

```python
for seed in range(80, 90):
    t0 = time.time()
    data = np.load('./data/H1L1_D0-35_256x128_7200s_4c_traindata_n1000_seed{}.npz'.format(seed), allow_pickle=True)
    noisy_image = data['noisy_image']
    signal_mask = data['signal_mask']
    clean_image = data['clean_image']
    f0 = data['f0']
    params = data['params']

    for i in range(clean_image.shape[0]):
        clean_image[i] = normalize(clean_image[i])

    for key in noisy_image.item().keys():
        n = noisy_image.item()[key].shape[0]
        for i in range(n):
            noisy_image.item()[key][i] = normalize(noisy_image.item()[key][i])


    print(time.time()-t0)
    np.savez("./data/H1L1_D0-35_256x128_7200s_4c_traindata_n1000_seed{}_norm.npz".format(seed), 
             noisy_image=noisy_image, signal_mask=signal_mask, clean_image=clean_image, f0=f0, params=params)
```
