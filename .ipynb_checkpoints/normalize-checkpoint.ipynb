```python
freq_nomial = 500
D = 35
Tsft = 14400
size = (512,64)
n = 400
# Initialize lists to store data from all seeds
all_noisy_images = {}  # Dictionary to store noisy images for each key
all_signal_masks = []
all_clean_images = []
all_f0 = []
all_params = []

for seed in range(1):
    t0 = time.time()

    # Load data
    file_path = f'./data/validation/{freq_nomial}Hz_H1L1_D0-{D}_{size[0]}x{size[1]}_{Tsft}s_4c_traindata_n{n}_seed{seed}_2323.npz'
    try:
        data = np.load(file_path, allow_pickle=True)
    except FileNotFoundError:
        print(f"Error: File {file_path} not found. Skipping seed {seed}.")
        continue

    noisy_image = data['noisy_image']
    signal_mask = data['signal_mask']
    clean_image = data['clean_image']
    f0 = data['f0']
    params = data['params']

    # Validate shapes
    expected_shape = (400, 256, 128, 4)
    if clean_image.shape != expected_shape:
        print(f"Warning: clean_image for seed {seed} has shape {clean_image.shape}, expected {expected_shape}")

    # Normalize clean_image (vectorized)
    clean_image = np.array([normalize(clean_image[i]) for i in range(clean_image.shape[0])])
    # Alternative vectorized normalization (if normalize supports it):
    # clean_image = normalize(clean_image)  # Uncomment if normalize can handle (100, 256, 128, 4)

    # Normalize noisy_image (dictionary-like)
    noisy_image_dict = noisy_image.item()
    for key in noisy_image_dict:
        n = noisy_image_dict[key].shape[0]
        if key not in all_noisy_images:
            all_noisy_images[key] = []
        # Normalize each sample
        normalized = np.array([normalize(noisy_image_dict[key][i]) for i in range(n)])
        all_noisy_images[key].append(normalized)

    # Append data
    all_signal_masks.append(signal_mask)
    all_clean_images.append(clean_image)
    all_params.append(params)

    seed_time = time.time() - t0
    print(f"Processed seed {seed} in {seed_time:.2f} seconds")

# Combine data
combined_clean_image = np.concatenate(all_clean_images, axis=0)  # Shape: (400, 256, 128, 4)
combined_signal_mask = np.concatenate(all_signal_masks, axis=0)  # Shape: (400, ...)
combined_params = np.concatenate(all_params, axis=0)  # Shape: (400, ...) or list

# Combine noisy_image dictionary
combined_noisy_image = {}
for key in all_noisy_images:
    combined_noisy_image[key] = np.concatenate(all_noisy_images[key], axis=0)

# Save combined data
output_file = f'./data/validation/{freq_nomial}Hz_H1L1_D0-{D}_{size[0]}x{size[1]}_{Tsft}s_4c_traindata_n{n}_seed2323_norm.npz'
np.savez(output_file,
         noisy_image=combined_noisy_image,
         signal_mask=combined_signal_mask,
         clean_image=combined_clean_image,
         f0=f0,
         params=combined_params)

print(f"Saved combined data to {output_file}")
```
